# Peptide → FASTA Converter

Converts space/newline-delimited peptide sequences into FASTA format.

**How to use:**
1. Run the first code cell to define the converter function.
2. Run the second cell — it will pop up a text box for you to paste your sequences into.
3. Run the third cell to preview and download the resulting `.fasta` file.

```
import re

WRAP_WIDTH = 60  # standard FASTA line wrap length

def peptides_to_fasta(text, prefix="Seq"):
    """
    Takes raw text with space/newline-delimited peptide sequences and
    returns a FASTA-formatted string.

    Optional custom headers: a token like 'Name:SEQUENCE' uses 'Name'
    as the header instead of an auto-generated one.
    """
    tokens = text.split()
    fasta_lines = []

    for i, token in enumerate(tokens, start=1):
        if ":" in token:
            name, seq = token.split(":", 1)
        else:
            name, seq = f"{prefix}_{i}", token

        seq = re.sub(r"[^A-Za-z\-\*]", "", seq).upper()
        if not seq:
            continue

        fasta_lines.append(f">{name}")
        for j in range(0, len(seq), WRAP_WIDTH):
            fasta_lines.append(seq[j:j + WRAP_WIDTH])

    return "\n".join(fasta_lines) + "\n"

print("Converter function ready.")
```

## Paste your sequences
Run this cell, then paste your peptide list into the text box that appears and press Enter (or click outside the box).

```
import ipywidgets as widgets
from IPython.display import display

text_area = widgets.Textarea(
    placeholder='Paste your peptide sequences here (space and/or newline separated)...',
    layout=widgets.Layout(width='100%', height='300px')
)
display(text_area)
```

## Convert and download
Run this cell after pasting your sequences above. It will show a preview and download `peptides.fasta`.

```
raw_text = text_area.value

if not raw_text.strip():
    print("No sequences found — make sure you pasted text into the box above and re-run this cell.")
else:
    fasta_output = peptides_to_fasta(raw_text)

    print("--- Preview (first 500 characters) ---\n")
    print(fasta_output[:500])

    with open('peptides.fasta', 'w') as f:
        f.write(fasta_output)

    from google.colab import files
    files.download('peptides.fasta')
```

## Alternative: upload a .txt file instead of pasting
If you'd rather upload a text file containing your sequences, run this cell instead.

```
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]
raw_text = uploaded[filename].decode('utf-8')

fasta_output = peptides_to_fasta(raw_text)

print("--- Preview (first 500 characters) ---\n")
print(fasta_output[:500])

with open('peptides.fasta', 'w') as f:
    f.write(fasta_output)

files.download('peptides.fasta')
```
